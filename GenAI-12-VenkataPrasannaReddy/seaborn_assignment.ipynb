```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("ipl_2008_2024_complete.csv")
df=df[df.season.isin([2020.0,2021.0,2022.0,2023.0,2024.0])] # for clear insights just taking only 5 season of data
df.info()
```

```python
df.season=df.season.astype("int32")
```

```python
cols=["batting_team","bowling_team","venue","city","toss_decision","season"]
df[cols]=df[cols].astype("category")
```

```python
df.sample(5)
```

```python
df.batting_team=df.batting_team.replace({"Kings XI Punjab":"Punjab Kings","Royal Challengers Bengaluru":"Royal Challengers Bangalore"})
df.bowling_team=df.bowling_team.replace({"Kings XI Punjab":"Punjab Kings","Royal Challengers Bengaluru":"Royal Challengers Bangalore"})
data=df.pivot_table(index="batting_team",columns="season",values="total_runs",aggfunc="sum")
```

## Task - 8 Multi-Plots & Figure-Level Plots

```python
df.batting_team=df.batting_team.cat.remove_unused_categories()
df.bowling_team=df.bowling_team.cat.remove_unused_categories()
g=sns.FacetGrid(data=df,row="batting_team")
g.map_dataframe(sns.regplot,x="boundaries_4",y="total_runs")
plt.show()
```
