## Task - 1 Safe Division Utility

```python
numerator=input("please provide numerator : ")
denominator=input("Please provide denominator : ")
try:
    result=int(numerator)/int(denominator)
except ValueError as ex1:
    print(ex1)
except ZeroDivisionError as ex2:
    print(ex2)
else:
    print(f"{numerator}/{denominator}={result}")
finally:
    print("Operation Complete")
```

## Task - 2 Bill Calculator with Error Handling

```python
prices=[120,350,'abc',500,-200,800]
total_price=0
for i in prices:
    try:
        if i>=0:
           total_price+=i
        else:
            raise ValueError(f"Negative price not allowed")
    except ValueError as ve:
        print(ve)
    except TypeError as te:
        print("Strings are not allowed")
    else:
        print(total_price)
```

## Task - 3 Custom Exception : Age Validator

```python
def check_age(age):
    if age<=0 or age>120:
        raise ValueError
    else:
        return "Valid Age"
try:
    age=int(input("Please enter your age"))
    result=check_age(age)
except ValueError as verr:
    print("Age must be between 1 and 120")
else:
    print(result)
finally:
    print("Operation complete")
```

## Task - 4 File Reader with Exceptional Handling

```python
filename=input("Please enter the filename with extension")
try:
    with open(filename,'r') as f:
        content=f.readlines()
except FileNotFoundError as fnfe:
    print(fnfe)
except PermissionError as pe:
    print(pe)
else:
    [print(i) for i in content[:3]]
finally:
    print("File operation attempted.")
```

## Task - 5 Mini Program : Safe Shopping Cart

```python
cart=[]
while(True):
    user_input=input("Please enter the price of the product or enter q to exit")
    if user_input=='q':
        break
    else:
        try:
            price=float(user_input)
            if price<0:
                raise ValueError("Price can't be negative.")
        except ValueError as verror:
            print(verror)
        else:
            cart.append(price)
print(f"Total no of items : {len(cart)}\nTotal bill : {sum(cart)}")
```
