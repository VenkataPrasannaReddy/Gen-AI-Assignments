## Task - 1 Basic Function 

```python
def apply_discount(price,discount_percent=5):
    if(discount_percent<=60):
        return price*(1-discount_percent/100)
    else:
        return "Discount not possible"
print(apply_discount(1000,10))
print(apply_discount(500))
```

## Task - 2 Recursive Function

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    elif n<0:
        return "Please provide a whole number"
    else:
        return n*factorial(n-1)
print(factorial(5))
print(factorial(0))
print(factorial(-3))
```

## Task - 3 Lambda Function

```python
price_with_gst=lambda x:x*1.18
print(price_with_gst(100))
price_with_gst_discount=lambda x:x*1.18*0.95 #considering discount as 5%
print(price_with_gst_discount(100))
```

## Task - 4 Using Map

```python
prices=[100,250,400,1200,50]
prices_with_gst=list(map(lambda x:x*1.18,prices))
print(prices,prices_with_gst,sep="\n")
```

## Task - 5 Using Filter

```python
prices=[100,250,400,1200,50,2000,850]
q1=list(filter(lambda x:x>500,prices))
q2=list(filter(lambda x:x<=500,prices))
print(q1,q2,sep="\n")
```

## Task - 6 Combined Utility Function

```python
def process_prices(prices):
    discounted_prices=list(map(lambda x:x*0.9,prices))
    filtered_prices=list(filter(lambda x:x>300,discounted_prices))
    return discounted_prices,filtered_prices
process_prices([100,500,900,50,750])
```

## Task - 7 Mini Problem

```python
def add_price(prices_list,price):
    return prices_list.append(price)
def get_average_price(prices_list):
    return sum(prices_list)/len(prices_list)
def get_max_price(prices_list):
    return max(prices_list)
prices_list=[]
while(True):
    option=input("Please select an option from below\
    1-Add price\
    2-Show Average Price\
    3-Show Highest Price\
    Q-Quit")
    if option=='Q':
        break
    elif int(option)==1:
        add_price(prices_list,int(input("Enter the price that needs to be added : ")))
    elif int(option)==2:
        print(get_average_price(prices_list))
    elif int(option)==3:
        print(get_max_price(prices_list))
    else:
        print("Please provide a valid option.")
```
