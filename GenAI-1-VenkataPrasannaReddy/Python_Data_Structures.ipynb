## Task 1 - Product Collections (Lists & Tuples)


```python
products=["mobile","laptop","tablet","desktop","smartwatch","earbuds"]
sample_product=("mobile",45000,"electronics")
print(products[1],products[-1])
products.append("Camera")
products.append("Headphones")
print(products)
sample_product_list=list(sample_product)
sample_product_list[1]=40000
sample_product=tuple(sample_product_list)
print(sample_product)
```

## Task 2 - Categories (Sets)

```python
Categories=["Electronics","Fashion","Home Appliances","Books","Toys","Sports","Beauty","Automotive"]
Categories_set=set(Categories)
Categories_set.add("Automotive")
print(len(Categories))
print(len(Categories_set)) # As length of set is same as list hence we have not added any new category to the set.
print("Electronics" in Categories_set)
print(len(Categories_set)) # By default in set we can store only unique values, so the length of the set will be same as the unique categories in the set.
```

## Task 3 - Product Pricing (Dictionaries)


```python
price_dict={"mobile":45000,"laptop":59999.99,"tablet":30000,"desktop":79999.99,"smartwatch":15000,"earbuds":5000}
price_dict["Camera"]=24999.99
price_dict["smartwatch"]=24000
print(price_dict.pop("toys","Category not found"))
print(f"Maximum price: {max(price_dict.values())}, Minimum price: {min(price_dict.values())}")
```

## Task 4 - Combined Operations



```python
catalog=[("mobile",45000,"electronics"),("Shirts",2000,"fashion"),("Curtains",5000,"home appliances"),("bike accessories",100,"automotive"),("Cricket kit",1500,"sports"),("football",2000,"sports"),("laptop",80000,"electronics"),("car accessories",10000,"automotive"),("volleyball",2500,"sports")]
category_to_products=dict()
for tuple in catalog:
    if tuple[2] not in category_to_products:
        category_to_products[tuple[2]]=[tuple[0]]
    else:
        category_to_products[tuple[2]].append(tuple[0])
max=0
category=str()
for i,j in category_to_products.items():
    if len(j)>max:
        max=len(j)
        category=i
print(category_to_products[category])
```
