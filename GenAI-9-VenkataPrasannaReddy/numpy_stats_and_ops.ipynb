## Task - 1 Creating Numpy Arrays

```python
import numpy as np

array1=np.arange(1,10)
array2=np.arange(1,10).reshape((3,3))
array3=np.array([10,20,30,40,50])
print(array1.shape,array2.shape,array3.shape)
print(array1.dtype,array2.dtype,array3.dtype)
```

## Task - 2 Important Mathemetical Operations

```python
A=np.array([10,20,30,40])
B=np.array([1,2,3,4])
print(A+B)
print(A-B)
print(A*B)
print(A/B)
print(A**2)
print(np.add(A,B))
print(np.subtract(A,B))
print(np.multiply(A,B))
print(np.divide(A,B))
print(np.power(A,2))
```

## Task - 3 Important Numpy Mathematical Formulas

```python
values=np.array([2,4,6,8,10])
print(np.power(values,0.5))
print(np.exp(values))
print(np.log(values))
print(np.sum(values))
print(np.cumulative_sum(values))
```

## Task - 4 Aggregation Operations

```python
data=np.array([[10,20,30],[40,50,60],[70,80,90]])
print(np.sum(data,axis=1))
print(np.sum(data,axis=0))
print(np.min(data))
print(np.max(data))
print(np.mean(data))
```

## Task - 5 Statistical Operations 

```python
marks=np.array([78,85,90,66,72,88,95,60])
print(np.mean(marks))
print(np.median(marks))
print(np.var(marks))
print(np.std(marks))
print(np.min(marks))
print(np.max(marks))
print(f"Range({np.min(marks)}-{np.max(marks)})")
```

## Task - 6 Percentiles & Sorting

```python
print(np.sort(marks))
print(np.percentile(marks,25))
print(np.percentile(marks,50))
print(np.percentile(marks,75))
print(np.count_nonzero(marks>np.average(marks)))
```

## Task - 7 Mini Use Case

```python
sales=np.array([1200,1500,900,2000,1800,1700,1600])
print(np.sum(sales))
print(np.average(sales))
print(f"Highest sales is on day {np.where(sales==np.max(sales))[0]+1}")
print(f"Lowest sales is on day {np.where(sales==np.min(sales))[0]+1}")
print(np.std(sales))
print(np.where(sales>np.mean(sales))[0]+1)
```
