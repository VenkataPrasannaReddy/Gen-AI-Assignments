## Task - 1 Write Sales Record to a File

```python
sales=[1200,450,980,1500,3000]
sales=list(str(sales[i])+'\n' if i!=len(sales)-1 else str(sales[i]) for i in range(len(sales)))
f=open("sales_data.txt","w")
f.writelines(sales)
f.close()
f=open("sales_data.txt","r")
print(f.read())
```

```python
#In csv format
sales=[1200,450,980,1500,3000]
sales=list(str(sales[i])+',' if i!=len(sales)-1 else str(sales[i]) for i in range(len(sales)))
f=open("sales_data.txt","w")
f.writelines(sales)
f.close()
f=open("sales_data.txt","r")
print(f.read())
```

## Task - 2 Read File in Different Ways

```python
with open("sales_data.txt",'r') as f:
    print(f.read())
    print('*'*50)
    f.seek(0)
    print(f.readline().strip('\n'))
    print('*'*50)
    f.seek(0)
    sales_list=f.readlines()
    sales_list=[int(i.strip('\n')) for i in sales_list]
    print(sales_list)
```

## Task - 3 Append New Sales

```python
new_sales=['\n5000\n','2500\n','1700']
with open("sales_data.txt",'a') as f:
    f.writelines(new_sales)
```

```python
f=open("sales_data.txt",'r')
print(f.read())
#Total no of lines after appending
print(f.read())
f.close()
```

## Task -4 General Summary Report from File

```python
with open('sales_data.txt','r') as f:
    sales_list=f.readlines()
sales_list=[int(i.strip('\n')) for i in sales_list]
print(sales_list)
print("Total sales : ",sum(sales_list))
print("Highest sale : ",max(sales_list))
print("Lowest sale : ",min(sales_list))
print("Average sale : ",sum(sales_list)/len(sales_list))
```

## Task - 6 Read File Safely

```python
import os
filename=input("Please give the filename to open : ")
if(os.path.exists(filename)):
    with open(filename,'r') as f:
        print(f.read())
else:
    print("File not found. Please check the filename")
```

## Task - 7 Mini Project

```python
prices={"Mouse":500,"Keyboard":800,"Monitor":7000,"Pendrive":400,"Camera":5000}
discount=int(input("Please provide discount percentage"))
with open("Product_Pricing.txt",'w+') as f:
    f.writelines("Product | Original Price | Discounted Price\n")
    discounted_price=[]
    for k,v in prices.items():
        discounted_price.append(v*(discount/100))
        f.writelines(k+" | "+str(v)+" | "+str(v*(discount/100))+"\n")
    f.seek(0)
    data=f.read()
print(data)
print("Total Items : ",data.count("\n")-1)
print("Average Disocunted Price : ",sum(discounted_price)/len(discounted_price))

```
