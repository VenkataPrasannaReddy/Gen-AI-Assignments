## Task - 1 Basic Class & Object Creation

```python
class Product:
    def __init__(self,name,price,category):
        self.name=name
        self.price=price
        self.category=category
    def get_info(self):
        print(f"{self.name} product comes under {self.category} category at Rs.{self.price} price")
    def apply_discount(self,discount):
        return self.price*(1-discount/100)
laptop=Product("Laptop",80000,"Electronics")
washing_machine=Product("Washing Machine",25000,"Home Appliances")
laptop.get_info()
washing_machine.get_info()
print(laptop.apply_discount(12))
print(washing_machine.apply_discount(7))
```

## Task - 2 Constructor & Encapsulation

```python
class Product:
    def __init__(self,name,price,category):
        self.name=name
        self.__price=price
        self.category=category
    def get_info(self):
        print(f"{self.name} product comes under {self.category} category")
    def apply_discount(self,discount):
        return self.__price*(1-discount/100)
    def get_price(self):
        return self.__price
    def set_price(self,new_price):
        if new_price>0:
            self.__price=new_price
iphone17=Product("Iphone17",83000,"Mobiles")
iphone17.set_price(92000)
iphone17.get_price()
```

## Task - 3 Inheritance (Single-level)

```python
class ElectronicProduct(Product):
    def __init__(self,name,price,category,warranty_years):
        super().__init__(name,price,category)
        self.warranty_years=warranty_years
    def get_info(self):
        print(f"{self.name} product comes under {self.category} category which comes with {self.warranty_years} years warranty")
s25ultra=ElectronicProduct("S25 Ultra",88999,"Mobiles",5)
s25ultra.get_info()
print(f"S25 Ultra mobile price : {s25ultra.get_price()}")
print(f"S25 Ultra mobile price after discount : {s25ultra.apply_discount(15)}")
```

## Task - 4 Polymorphism (Method Overriding)

```python
class Laptop(Product):
    def __init__(self,name,price,category):
        super().__init__(name,price,category)
    def get_info(self):
        print(f"{self.name} product comes under {self.category} category which is a laptop")
        
class Mobile(Product):
    def __init__(self,name,price,category):
        super().__init__(name,price,category)
    def get_info(self):
        print(f"{self.name} product comes under {self.category} category which is a mobile")

lenovo_ideapad=Laptop("Lenovo Ideapad",75000,"Electronics")
dell_inspiron=Laptop("Dell Inspiron",65000,"Electronics")
samsung_galaxy=Mobile("Samsung Galaxy",45000,"Electronics")
iphone_14=Mobile("Iphone 14",90000,"Electronics")
list_of_products=[lenovo_ideapad,dell_inspiron,samsung_galaxy,iphone_14]
for product in list_of_products:
    product.get_info()  
```

## Task - 5 Abstraction (Using Abstract Base Class)

```python
from abc import ABC,abstractmethod

class Payment(ABC):
    @abstractmethod
    def process_payment(self):
        pass
class CreditCardPayment(Payment):
    def process_payment(self,amount):
        print(f"Credit payment of Rs.{amount} has been successfull.")
class UPIPayment(Payment):
    def process_payment(self,amount):
        print(f"UPI payment of Rs.{amount} has been successfull.")

creditcardpayment=CreditCardPayment()
creditcardpayment.process_payment(2000)
upipayment=UPIPayment()
upipayment.process_payment(5000)
```

## Task - 6 Magic Methods & Operator Overloading

```python
class Product:
    def __init__(self,name,price,category):
        self.name=name
        self.__price=price
        self.category=category
    def get_info(self):
        print(f"{self.name} product comes under {self.category} category")
    def apply_discount(self,discount):
        return self.__price*(1-discount/100)
    def get_price(self):
        return self.__price
    def set_price(self,new_price):
        if new_price>0:
            self.__price=new_price
    def __str__(self):
        return f"{self.__class__.__name__}({self.name},{self.__price},{self.category})"
    def __add__(self,other):
        return self.__price+other.get_price()
ac=Product("Blue Star AC",40000,"Home Appliances")
tv=Product("LG TV",45000,"Electronics")
print(ac)
print(ac+tv)
```

## Task - 7 Mini Project : Simple Inventory System

```python
class Product:
    def __init__(self,name,price,category):
        self.name=name
        self.__price=price
        self.category=category
    def get_info(self):
        print(f"{self.name} product comes under {self.category} category")
    def apply_discount(self,discount):
        return self.__price*(1-discount/100)
    def get_price(self):
        return self.__price
    def set_price(self,new_price):
        if new_price>0:
            self.__price=new_price
    def __add__(self,other):
        return self.__price+other.get_price()
        
class Inventory:
    def __init__(self,products=None):
        if products==None:
            self.products=[]
        else:
            self.products=products
    def add_product(self,product):
        self.products.append(product)
    def remove_product(self,name):
        self.products.remove(name)
    def get_total_value(self):
        sum=0
        for product in self.products:
            sum+=product.get_price()
        return sum
    def show_all_products(self):
        for product in self.products:
            product.get_info()


class Store:
    def __init__(self,store_name,inventory):
        self.store_name=store_name
        self.inventory=inventory
    def add_new_product(self,name,price,category):
        self.inventory.add_product(Product(name,price,category))
    def show_summary(self):
        self.inventory.show_all_products()
        print("Total value of all the items : ",self.inventory.get_total_value())

inventory=Inventory()
store=Store("Blinkit",inventory)
store.add_new_product("Rice",2500,"Food")
store.add_new_product("Chana Dal",180,"Food")
store.add_new_product("Shirt",1500,"Fashion")
store.show_summary()
p1=Product("Rice",2500,"Food")
p2=Product("Shirt",1500,"Fashion")
print(p1+p2)
```
