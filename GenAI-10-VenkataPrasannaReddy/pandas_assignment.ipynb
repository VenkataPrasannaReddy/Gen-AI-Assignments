## Task - 1 Pandas Series Basics

```python
import pandas as pd
```

```python
marks=[78,85,90,66,72]
pd_series=pd.Series(marks,name="marks")
print(pd_series)
print(pd_series.index)
print(pd_series.dtype)
print(pd_series[0])
print(pd_series.tail(2))
```

## Task - 2  Mathematical Operations on Series

```python
pd_series+=5 # Adding 5 grace marks to all students
print(pd_series)
pd_series-=2 # subtracting 2 marks from all students
print(pd_series)
pd_series*=1.05 # multiplying  all marks by 1.05
print(pd_series)
pd_series/=2
print(pd_series)
```

## Task - 3 Python Functionalities on Series

```python

pd_series=pd.Series(marks,name="marks")
print(pd_series.max())
print(pd_series.min())
print(pd_series.sum())
print(pd_series.mean())
passed_students=lambda x:x>=70
print(pd_series[passed_students].size) # no of students passed
```

## Task - 4 Create a DataFrame

```python
students={
    'Name':['Amit','Neha','Rahul','Sneha','Pooja'],
    'Marks':[78,85,90,66,72],
    'Subject':["Math",'Math','Science','Science','Math']
}
df_students=pd.DataFrame(students)
print(df_students.head(3))
print(df_students.tail(2))
print(df_students.shape)
print(df_students.columns)
```

## Task - 9 Mini Use Case : Sales Data Analysis

```python
sales={"Day":['Mon','Tue','Wed','Thu','Fri'],'Revenue':[1200,1500,900,2000,1800]}
df_sales=pd.DataFrame(sales)
df_sales
```

```python
#Total Revenue
print(df_sales["Revenue"].sum())
```

```python
print(df_sales["Revenue"].mean())
```

```python
print(df_sales[df_sales["Revenue"].max()==df_sales["Revenue"]]["Day"])
```

```python
print(df_sales[df_sales["Revenue"]>df_sales["Revenue"].mean()])
```

```python
df_sales.plot(x="Day",y="Revenue",kind="bar")
```
